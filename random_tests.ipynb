numpy array torch tensor

```python
import numpy as np

#data from numpy -> tensor: torch.from_numpy(array)
#tensor -> array: torch.Tensor.numpy()

array= np.arange(1.0, 11.0)
print('array:\n', array)

tensor= torch.from_numpy(array).type(torch.float32)
print('new tensor\n', tensor)

#to numpy array:

to_array= torch.Tensor.numpy(tensor)
print('new array:\n', to_array)
```

Reproduceaility

```python
RANDOMSEED= 42

torch.manual_seed(RANDOMSEED)
tensor= torch.rand(1,9).reshape(1,3,3) # random
#Muste set seed each time using method
torch.manual_seed(RANDOMSEED)
tensor1= torch.rand(1,9).reshape(1,3,3) # random
print(tensor== tensor1)
```
