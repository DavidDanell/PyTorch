Manipulating tensors (tensor operations) 

```python
# Create a tensor of values and add a number to it
tensor = torch.tensor([1, 2, 3])
print(tensor + 10)

print(tensor * 10) #tensor values dont change unless they are assigned tensor= tensor *3
tensor = tensor - 10
print(tensor)
tensor = tensor + 10
print(tensor)

#built in functions:

tensor= torch.add(tensor, 10)
print(tensor)
tensor= torch.sub(tensor, 10)
#torch.mul
#more common to use operator * 
# Element-wise multiplication (each element multiplies its equivalent, index 0->0, 1->1, 2->2)
print(tensor, "*", tensor)
print("Equals:", tensor * tensor)

# MATRIX MULTIPLICATION (THE MOST COMMON OPERATION) torch.matmul()

# RULES FOR MATRIX MUL: 
#1. INNER DIMENSIONS MUST MATCH:

#(3, 2) @ (3, 3) WONT WORK
# (3,3) @ (3,3) WILL WORK

#3. THE RESULT OF MATIX MUL WILL BE THE SHAPE OF THE OUTER DIMENSIONS:

# (1, 3) @ (3, 1) = (1, 1), @ =  PYTHONS MATRIX MUL

#element wise:
print('\n',tensor * tensor)


#matrix mul
print(tensor.matmul(tensor))

```

MOST COMMON ERRORS IN DEEP LEARNNG (SHAPE ERRORS)

```python
# Shapes need to be in the right way  
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10],
                         [8, 11], 
                         [9, 12]], dtype=torch.float32)


print(tensor_A.shape, tensor_B.shape)
torch.matmul(tensor_A, tensor_B) # (this will error)
```

```python
tensor_B.T   #.T = transpose 1 solution
                       # torch.transpose(tensor_B, dim0, dim1) 2 solution

print(torch.transpose(tensor_B, 0, 1)) #troanspose ändrar struktur på matrisen, byter plats? (T)

print(torch.mm(tensor_A, tensor_B))

```

```python
# Since the linear layer starts with a random weights matrix, let's make it reproducible (more on this later)
tensor_A=tensor_A.T

torch.manual_seed(42)
# This uses matrix multiplication
linear = torch.nn.Linear(in_features=3, # in_features = matches inner dimension of input 
                         out_features=6) # out_features = describes outer value 
x = tensor_A
output = linear(x)
print(f"Input shape: {x.shape}\n")
print(f"Output:\n{output}\n\nOutput shape: {output.shape}")

```


Reshaping, stacking, squeezing and unsqueezing

```python
tensor= torch.arange(0, 10)
print(tensor.shape)
print(tensor)

#extra dimension

tensor=tensor.reshape(1, 10)

print(tensor.shape)
print(tensor)


x=tensor
# Stack tensors on top of each other
x_stacked = torch.stack([x, x, x, x, x], dim=0) # try changing dim to dim=1 and see what happens
#x_stacked


print(f"Previous tensor: {x}")
print(f"Previous shape: {x.shape}")
# Remove extra dimension from x_reshaped
x_squeezed = x.squeeze()
print(f"\nNew tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")


print(f"Previous tensor: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

## Add an extra dimension with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```
